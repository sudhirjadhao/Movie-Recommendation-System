```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
movies = pd.read_csv("/content/Movies_dataset.csv")
movies.head()
```

```python
movies.columns
```

```python
movies.info()
```

```python
movies['tags'] =movies['overview'] + movies['genre']
movies.head()
```

```python
movies_new =movies[['id', 'title', 'tags']]
```

```python
movies_new.head()
```

```python
#converts text into a bag-of-words (BoW) model, keep top 100 words and remove english stop words
cv = CountVectorizer(max_features=1000, stop_words='english')
```

```python
cv
```

```python
##learn vocabulary from data converted to array, Unicode format ('U'), ensuring text compatibility
vec = cv.fit_transform(movies_new['tags'].values.astype('U')).toarray()
```

```python
vec
```

```python
vec.shape
```

```python
#Computes cosine similarity between all movie vectors.#Values range from 0 (no similarity) to 1 (identical text content)
sim = cosine_similarity(vec)
```

```python
sim
```

```python
movies_new[movies_new['title'] == 'The Shawshank Redemption']
```

```python
dist = sorted(list(enumerate(sim[0])), reverse=True, key=lambda vec:vec[1])
```

```python
dist
```

```python
## top 5 movies related to Shawshank Redemption
for i in dist[0:5]:
  print(movies_new.iloc[i[0]].title)
```

```python
# Function to recommend to 5 related movies
def recommend(movie):
  index = movies_new[movies_new['title'] == movie].index[0]
  distance = sorted(list(enumerate(sim[index])), reverse=True, key=lambda vec:vec[1])
  for i in distance[0:5]:
    print(movies_new.iloc[i[0]].title)
```

```python
recommend('Iron Man')
```
